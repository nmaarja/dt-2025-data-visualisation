```python
sentence = "racecarlevelfield"

# extract first 3 letters
print("Print first 3 letters:", sentence [0:3]) #starts: ends
print (sentence [:3])

#letters from 5-9
print (sentence[4:9])

#letters from 5th to end
print (sentence [5:])

#extract the last character
print (sentence[-1])

# extract last 4 characters
print (sentence[-4:])

#extract first letter of the substring starting from 5
#sentence[5:] will extract a substring starting from 5
#[0] this will take the first character of that substring
print (sentence[5:][0])

#extract first 5 letters of the substring starting from 5
print (sentence[5:][:5])

#exract entire text
print (sentence[:])

#extract every 2nd character
print(sentence [::2])
```

```python
sentence = "Alice went to the park and saw a beautiful bird singing"

words = sentence.split(" ") #split the sentence to words
print(words)

#extract first word
print(words[0])

# first 3 words
print (*words[0:3])

# last 3 words
print(*words[-3:])

#print number of words
print(len(words))
```
