```python
sentence = "The little girl is playing with her friends in the park after school"

# extract first 10 letters
print(sentence[:10])

# extract last 6 letters
print(sentence[-6:])

#extract first 8 letters of the substring starting from 10
print (sentence[10:][:8])

words = sentence.split(" ") #split the sentence to words
print(words)

#extract words that are longer than 4 letters
words = sentence.split()                
long_sentence = [w for w in words if len(w) > 4]
print(long_sentence)

#print number of words
print(len(words))

# first 5 words
print (*words[0:5])


```

```python
from wordcloud import WordCloud
from PIL import Image # import image class from PIL
import urllib.request # library to work with URL
import numpy as np # in order to work with image masks

text = urllib.request.urlopen("https://raw.githubusercontent.com/nmaarja/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/class_work/text.txt").read().decode()


img =urllib.request.urlretrieve("https://drive.google.com/uc?export=download&id=1N3Jtk_Swwws1r1aufIlrfR4BLG5dBPKE",
    "image1.png")

# using this shape as a mask, change baackround color and add stopwords
WordCloud(mask=np.array(Image.open("image1.png")),background_color="pink",stopwords=["ja", "kui", "et", "pigem", "aga", "on"]).generate_from_text(text).to_image()
```

I can use wordcloud to see which words have used the most.
I did this wordcloud about blue led lights. And as expected the most used words are "led", "valgust" which means "light", "rohkem" which means "more" and "sinine" which means "blue".
