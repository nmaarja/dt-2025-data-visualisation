```python
sentence = "The little girl is playing with her friends in the park after school"

# extract first 10 letters
print(sentence[:10])

# extract last 6 letters
print(sentence[-6:])

#extract first 8 letters of the substring starting from 10
print (sentence[10:][:8])

words = sentence.split(" ") #split the sentence to words
print(words)

#extract words that are longer than 4 letters
words = sentence.split()                
long_sentence = [w for w in words if len(w) > 4]
print(long_sentence)

#print number of words
print(len(words))

# first 5 words
print (*words[0:5])


```

```python
from wordcloud import WordCloud
from PIL import Image # import image class from PIL
import urllib.request # library to work with URL
import numpy as np # in order to work with image masks

text = urllib.request.urlopen("https://drive.google.com/uc?export=download&id=1MBQ5OebeGeHrKq5ixlL0wO-noZo-HhKFbKqT0qx23Qs").read().decode()
# https://docs.google.com/document/d/1MBQ5OebeGeHrKq5ixlL0wO-noZo-HhKFbKqT0qx23Qs/edit?usp=sharing

img = urllib.request.urlretrieve("https://drive.google.com/uc?export=download&id=1m8GdDDAbDAJAXQQoRr-AYoNQkxjlEJBI",
    "image1.png")

WordCloud(mask=np.array(Image.open("image1.png")),background_color="gray").generate_from_text(text).to_image()
```
